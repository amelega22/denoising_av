```python
import obspy
from obspy import UTCDateTime
from obspy.clients.fdsn import Client
```

```python
# params (from David)

# IRIS is now EARTHSCOPE, either work though
SOURCE = 'EARTHSCOPE'
NETWORK = 'AV'
STATION = 'SSLS'
LOCATION = ''
CHANNEL = 'BH*'

START = UTCDateTime('2025-10-01T00:00:00')
# get one hour of data:
END = START + 3600 
```

```python
client = Client(SOURCE)

waveform_data = client.get_waveforms(NETWORK, STATION, LOCATION, CHANNEL, START, END)
# metadata = client.get_stations(network='AV', station=STATION, level='channel')
```
